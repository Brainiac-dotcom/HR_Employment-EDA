# HR EMPLOYEMENT ATTRITION EDA

```python
import pandas as pd
import numpy as np 

df =  pd.read_csv(r"C:\Users\DELL\Downloads\WA_Fn-UseC_-HR-Employee-Attrition.csv")
pd.set_option('display.max_columns', None)
df
```

```python
df.describe()
```

```python
# Number of employees who left the company vs those who stayed
Attrition_Count = df['Attrition'].value_counts()
Attrition_Count
```

```python
# Comparing numeric values by Attrition
(df.groupby('Attrition')[['Age','DailyRate','DistanceFromHome','Education','MonthlyRate','MonthlyIncome','HourlyRate','YearsSinceLastPromotion','YearsWithCurrManager','YearsInCurrentRole','YearsAtCompany','JobLevel','PercentSalaryHike']].value_counts().reset_index())
```

```python
# Check how many employee in a particular age grade left or stayed
with pd.option_context('display.max_rows', None):
    display(df.groupby('Attrition')['Age'].value_counts().reset_index())
```

```python
# Sum,Averaga & Income of MonthlyIncome & Attrition
Monthlyincome_Attrition = df.groupby('Attrition')['MonthlyIncome'].agg(['mean','sum','count'])
Monthlyincome_Attrition
```

```python
Averge_Monthlyincome_Attrition = df.groupby('Attrition')['MonthlyIncome'].mean().reset_index()
Averge_Monthlyincome_Attrition
```

# Comparing Categorical Varaibles with Attrition

```python
Department_Attrition = pd.crosstab(df['Department'],df['Attrition'])
Department_Attrition
```

```python
Gender_Attrition = pd.crosstab(df['Gender'],df['Attrition'])
Gender_Attrition
```

```python
# Average MonthlyIncome of JobRole with Attrition
Average_MonthlyIncome_JobRole_Attrition = df.pivot_table(index = 'JobRole', columns = 'Attrition', values = 'MonthlyIncome', aggfunc= 'mean')
Average_MonthlyIncome_JobRole_Attrition
```

```python
#Average YearsSinceLastPromotion of JobRole with Attrition
Average_YearsSinceLastPromotion_JobRole_Attrition = df.pivot_table(index = 'JobRole', columns = 'Attrition', values = 'YearsSinceLastPromotion', aggfunc= 'mean')
Average_YearsSinceLastPromotion_JobRole_Attrition 
```

```python
# Attrition in OverTime
Attrition_OverTime = df.groupby('Attrition')['OverTime'].value_counts().reset_index()
Attrition_OverTime
```

```python
Correlation = df.corr(numeric_only = True)
Correlation
```

# VISUALIZATION 

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# chart showing the number of employees who left the company vs those who stayed
sns.countplot(x = 'Attrition', data = df)
plt.title('Attrition Count')
plt.xlabel('Attrition')
plt.ylabel('Count')
plt.show()
```

```python
# Visual of Department vs Attrition
sns.countplot(x = 'Department', hue = 'Attrition', data = df)
plt.title('Department vs Attrition')  
plt.ylabel('Count')
plt.show()
```

```python
# chart showing Average MonthlyIncome of JobRole with Attrition
Average_MonthlyIncome_JobRole_Attrition.plot(kind = 'barh', figsize = (10,7))
plt.title('Average Monthly Income by Job Role and Attrition')
plt.xlabel('Average Monthly Income')
plt.ylabel('Job Role')
plt.show()
```

```python
sns.countplot(x = 'OverTime', hue = 'Attrition', data = df)
plt.title('OverTime vs Attrition')  
plt.ylabel(' ')
plt.show()
```

```python
# Visualization on Correlation of the dataset
Correlation = df.drop(columns=['EmployeeCount','EmployeeNumber','StandardHours']).corr(numeric_only = True)
plt.figure(figsize=(16, 12))
sns.heatmap(Correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.show()
```

```python
# Chart Average YearsSinceLastPromotion of JobRole with Attrition
Average_YearsSinceLastPromotion_JobRole_Attrition.plot(kind = 'barh', figsize = (10,7))
plt.title('Average Years Since Last Promotion by Job Role and Attrition')   
plt.xlabel('Average Years Since Last Promotion')
plt.ylabel('Job Role')
plt.show()
```

```python
# Visualization of Averaga  of MonthlyIncome & Attrition
Averge_Monthlyincome_Attrition.plot(kind='bar',
    color=['skyblue', 'salmon'],
    figsize=(6,4)
)

plt.title('Average Monthly Income by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Average Monthly Income')
plt.xticks(rotation=0)
plt.show()
```
